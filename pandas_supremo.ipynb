```python
dataframe_vendas_analise = pandas.read_csv("./vendas_tech.csv", low_memory=False)
dataframe_vendas_analise = dataframe_vendas_analise.drop(columns=["Data_Base"])
dataframe_vendas_analise = dataframe_vendas_analise.drop_duplicates()
dataframe_vendas_analise["Data"] = pandas.to_datetime(dataframe_vendas_analise["Data"], format="%Y-%m-%d")

# Encadeia várias operações para uma mesma coluna
dataframe_vendas_analise["Loja"] = dataframe_vendas_analise["Loja"].fillna("Online").str.strip().str.title()

# Cria uma nova coluna "Faturamento" para cada linha, atribuindo como valor a multiplicação da coluna "Qtd" pela coluna "Preco_Unitario"
dataframe_vendas_analise["Faturamento"] = dataframe_vendas_analise["Qtd"] * dataframe_vendas_analise["Preco_Unitario"]

dataframe_vendas_analise.head()
```

```python
regioes = {
	'São Paulo': "Sudeste", 
	'Belo Horizonte': "Sudeste",
	'Online': "Online", 
	'Rio De Janeiro': "Sudeste",
	'Salvador': "Nordeste",
	'Recife': "Nordeste",
	'Curitiba': "Sul",
	'Porto Alegre': "Sul"
}

# Define uma função que recebe uma string loja e retorna a região de acordo com o dicionário de regiões
def definir_regiao(loja: str):
	return regioes[loja]

# Mapeia a coluna "Loja" da base de dados mapeando os valores dela utilizando a função definir_regiao.
# Adiciona uma coluna "Regiao" que terá o nome da região de acordo com a coluna "Loja"
dataframe_vendas_analise["Regiao"] = dataframe_vendas_analise["Loja"].map(definir_regiao)

# Outra forma de adicionar a coluna mapeando pela coluna "Loja" é passar como argumento o dicionário ao invés de uma função
dataframe_vendas_analise["Regiao"] = dataframe_vendas_analise["Loja"].map(regioes)

dataframe_vendas_analise.sample(10)
```

### Mesclando Bases de Dados

```python
# Desafio: Quais gerentes bateram a meta em janeiro de 2023?

dataframe_vendas = pandas.read_csv("./vendas_tech.csv", low_memory=False)
dataframe_gerentes = pandas.read_excel("./gerentes_lojas.xlsx")

dataframe_vendas = dataframe_vendas.rename(columns={"Qtd": "Quantidade"})
dataframe_vendas = dataframe_vendas.drop_duplicates(subset=["ID_Pedido"])
dataframe_vendas = dataframe_vendas[["Data", "Loja", "Preco_Unitario", "Quantidade"]]
dataframe_vendas["Loja"] = dataframe_vendas["Loja"].fillna("Online")
dataframe_vendas["Loja"] = dataframe_vendas["Loja"].str.strip().str.title()

dataframe_vendas["Data"] = pandas.to_datetime(
    dataframe_vendas["Data"], format="%Y-%m-%d"
)
dataframe_vendas = dataframe_vendas[
    (dataframe_vendas["Data"].dt.year == 2023)
    & (dataframe_vendas["Data"].dt.month == 1)
]

dataframe_vendas["Faturamento"] = (
    dataframe_vendas["Quantidade"] * dataframe_vendas["Preco_Unitario"]
)

dataframe_faturamento = dataframe_vendas[["Loja", "Faturamento"]]
dataframe_faturamento = dataframe_faturamento.groupby("Loja").sum()
dataframe_faturamento = dataframe_faturamento.reset_index()

dataframe_gerentes["Loja"] = dataframe_gerentes["Loja"].str.strip().str.title()
dataframe_gerentes["Gerente"] = dataframe_gerentes["Gerente"].str.strip().str.title()

# Mescla as bases de dados de acordo com o valor da coluna "Loja". As duas bases de dados serão unidas adicionando as colunas de "dataframe_gerentes" à "dataframe_faturamento_metas_gerentes". Os valores de uma linha do dataframe de gerentes serão incluídos nas linhas do dataframe de faturamento quando o valor da coluna "Loja" de gerentes for igual ao valor da coluna "Loja" na linha do dataframe de faturamento
dataframe_faturamento_metas_gerentes = dataframe_faturamento.merge(
    dataframe_gerentes, on="Loja", how="left"
)


"""
- O argumento "left_on" define a coluna a ser usada para mesclar no dataframe da esquerda.
- O argumento "right_on" define a coluna a ser usada para mesclar no dataframe da direita.
- O dataframe da esquerda é sempre a que chama o método "merge".
- O dataframe da direita é sempre a que é passada como parâmetro para o método merge.
- Por padrão, o pandas realiza um inner join. Ou seja, o dataframe de resultado terá linhas apenas para os valores que existirem em ambos os dataframes nas colunas indicadas
- No exemplo abaixo, as linhas do dataframe de faturamento serão mescladas com as linhas do dataframe de gerentes quando o valor da coluna "Loja" no dataframe de faturamento for igual ao valor da coluna "Cidade" no dataframe de gerentes.
dataframe_faturamento_metas_gerentes = dataframe_faturamento.merge(
  	dataframe_gerentes, left_on="Loja", right_on="Cidade"
)
"""

"""
- Para definir o tipo de join, utilizamos o parâmetro "how". Na maioria dos casos de uso fazemos left joins, right joins e inner joins. 
- Left join vai garantir que todos os valores da coluna especificada do dataframe da esquerda existam no dataframe de resultado. 
- Right join vai garantir que todos os valores da coluna especificada do dataframe da direita existam no dataframe de resultado.
- Inner join vai garantir que a coluna de resultado só contenha valores que existam ao mesmo tempo no dataframe da esquerda e no dataframe da direita.
dataframe_faturamento_metas_gerentes = dataframe_faturamento.merge(
  	dataframe_gerentes, left_on="Loja", how="left"
)
"""

dataframe_faturamento_metas_gerentes["Meta_Atingida"] = numpy.where(
    dataframe_faturamento_metas_gerentes["Faturamento"]
    >= dataframe_faturamento_metas_gerentes["Meta_Mensal"],
    "Sim",
    "Não",
)

dataframe_faturamento_metas_gerentes.to_csv(
    "desafios/gerentes_bateram_meta_jan_2023.csv"
)

print(dataframe_faturamento_metas_gerentes.head(n=10))
```
